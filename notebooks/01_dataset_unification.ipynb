```python
# Cell 4: Compare Dataset Schemas
# Purpose: Identify common features and differences across datasets

print("🔍 Dataset Schema Comparison")
print("=" * 60)

# Get column sets
cols_2018 = set(df_2018.columns)
cols_unsw = set(df_unsw.columns)
cols_2017 = set(df_2017.columns)

print("\n📊 Total Columns per Dataset:")
print(f"   - CSE-CIC-IDS-2018: {len(cols_2018)} columns")
print(f"   - UNSW-NB15: {len(cols_unsw)} columns")
print(f"   - CIC-IDS-2017: {len(cols_2017)} columns")

# Common columns across all three datasets
common_all = cols_2018 & cols_unsw & cols_2017
print(f"\n✅ Common in ALL 3 datasets: {len(common_all)} columns")
if common_all:
    print(f"   {sorted(common_all)}")

# Common between CIC-IDS datasets (2017 and 2018)
common_cic = cols_2018 & cols_2017
print(f"\n🔗 Common in CIC-IDS (2017 & 2018): {len(common_cic)} columns")
print(f"   {sorted(common_cic)}")

# Unique to each dataset
unique_2018 = cols_2018 - cols_2017 - cols_unsw
unique_2017 = cols_2017 - cols_2018 - cols_unsw
unique_unsw = cols_unsw - cols_2018 - cols_2017

print(f"\n🔸 Unique to CSE-CIC-IDS-2018: {len(unique_2018)} columns")
print(f"   {sorted(unique_2018)}")

print(f"\n🔸 Unique to CIC-IDS-2017: {len(unique_2017)} columns")
print(f"   {sorted(unique_2017)}")

print(f"\n🔸 Unique to UNSW-NB15: {len(unique_unsw)} columns")
print(f"   {sorted(unique_unsw)}")

# Label column analysis
print("\n\n🏷️  Label Column Analysis:")
print(f"   CSE-CIC-IDS-2018: 'Label' - {df_2018['Label'].nunique()} classes")
print(
    f"   UNSW-NB15: 'Label' (binary) + 'attack_cat' - {df_unsw['attack_cat'].nunique()} classes")
print(
    f"   CIC-IDS-2017: 'ClassLabel' - {df_2017['ClassLabel'].nunique()} classes")
```

```python
# Cell 5: Define Unified Feature Set
# Purpose: Identify which columns to use in unified dataset

print("🔧 Defining Unified CIC Feature Set")
print("=" * 60)

# Common features between 2017 and 2018 (excluding label columns)
common_features = sorted(list(cols_2017 & cols_2018 - {'Label', 'ClassLabel'}))

# Additional features from 2018 (will be NaN for 2017 rows)
additional_2018_features = sorted(list(cols_2018 - cols_2017 - {'Label'}))

# All features for unified dataset
all_features = common_features + additional_2018_features

print("\n📊 Feature Breakdown:")
print(f"   ✅ Common features (2017 & 2018): {len(common_features)}")
print(f"   ➕ Additional from 2018: {len(additional_2018_features)}")
print(f"   📦 Total features: {len(all_features)}")

print(f"\n✅ Common Features ({len(common_features)}):")
for i, feat in enumerate(common_features, 1):
    print(f"   {i:2d}. {feat}")

print(f"\n➕ Additional 2018 Features ({len(additional_2018_features)}):")
for i, feat in enumerate(additional_2018_features, 1):
    print(f"   {i:2d}. {feat}")

print("\n💡 Note: These 20 additional features will be:")
print("   - Present for 2018 rows (6.66M samples)")
print("   - NaN for 2017 rows (9.17M samples)")
print("   - Will be imputed using appropriate strategy")

# Save feature list for later use
feature_list = {
    'common_features': common_features,
    'additional_2018_features': additional_2018_features,
    'all_features': all_features
}

print("\n✅ Feature list saved to memory")
```

```python
# Cell 6: Label Mapping Strategy
# Purpose: Map 2018's 15 classes to 2017's 8 coarse classes for unified taxonomy

print("🏷️  Label Taxonomy Unification")
print("=" * 60)

# Show current label distributions
print("\n📊 Current Label Distributions:")
print("\n2017 ClassLabel (8 classes):")
print(df_2017['ClassLabel'].value_counts().sort_index())

print("\n2018 Label (15 classes):")
print(df_2018['Label'].value_counts().sort_index())

# Define mapping: 2018 detailed → 2017 coarse
label_mapping_2018_to_2017 = {
    # Benign
    'Benign': 'Benign',

    # DDoS variants → DDoS
    'DDoS attacks-LOIC-HTTP': 'DDoS',
    'DDOS attack-HOIC': 'DDoS',
    'DDOS attack-LOIC-UDP': 'DDoS',

    # DoS variants → DoS
    'DoS attacks-Hulk': 'DoS',
    'DoS attacks-GoldenEye': 'DoS',
    'DoS attacks-Slowloris': 'DoS',
    'DoS attacks-SlowHTTPTest': 'DoS',

    # Bot → Botnet
    'Bot': 'Botnet',

    # Infiltration (typo in 2018: "Infilteration")
    'Infilteration': 'Infiltration',

    # Bruteforce variants → Bruteforce
    'SSH-Bruteforce': 'Bruteforce',
    'FTP-BruteForce': 'Bruteforce',

    # Web attacks → Webattack
    'Brute Force -Web': 'Webattack',
    'Brute Force -XSS': 'Webattack',
    'SQL Injection': 'Webattack'
}

print("\n🔗 Label Mapping (2018 → 2017):")
print("=" * 60)
for old_label, new_label in sorted(label_mapping_2018_to_2017.items()):
    print(f"   {old_label:30s} → {new_label}")

# Apply mapping to 2018 dataset (create new column, don't modify original)
df_2018['UnifiedLabel'] = df_2018['Label'].map(label_mapping_2018_to_2017)

# Check for any unmapped labels
unmapped = df_2018[df_2018['UnifiedLabel'].isna()]['Label'].unique()
if len(unmapped) > 0:
    print(f"\n⚠️  WARNING: Unmapped labels found: {unmapped}")
else:
    print("\n✅ All 2018 labels successfully mapped")

# Show unified distribution for 2018
print("\n📊 2018 After Mapping (8 classes):")
print(df_2018['UnifiedLabel'].value_counts().sort_index())

# For 2017, just use ClassLabel as UnifiedLabel
df_2017['UnifiedLabel'] = df_2017['ClassLabel']

print("\n📊 2017 UnifiedLabel (already 8 classes):")
print(df_2017['UnifiedLabel'].value_counts().sort_index())

print("\n✅ Label unification complete")
print("\n🎯 Final Unified Taxonomy: 8 classes")
print(f"   {sorted(df_2017['UnifiedLabel'].unique())}")

# Save mapping for later use
label_mapping = label_mapping_2018_to_2017

print("\n✅ Label mapping saved to memory")
```

```python
# Cell 7: Combine Datasets
# Purpose: Merge 2017 and 2018 into unified dataset with 77 features + UnifiedLabel

print("🔗 Combining CIC-IDS Datasets")
print("=" * 60)

# Prepare 2018 dataset
print("\n📦 Preparing 2018 dataset...")
df_2018_prepared = df_2018[all_features + ['UnifiedLabel']].copy()
df_2018_prepared['Dataset_Source'] = '2018'
print(f"   Shape: {df_2018_prepared.shape}")
print(
    f"   Columns: {len(df_2018_prepared.columns)} (77 features + label + source)")

# Prepare 2017 dataset (need to add 20 missing columns as NaN)
print("\n📦 Preparing 2017 dataset...")
df_2017_prepared = df_2017[common_features + ['UnifiedLabel']].copy()

# Add 20 additional columns from 2018 as NaN
for col in additional_2018_features:
    df_2017_prepared[col] = np.nan

# Reorder columns to match 2018
df_2017_prepared = df_2017_prepared[all_features + ['UnifiedLabel']]
df_2017_prepared['Dataset_Source'] = '2017'
print(f"   Shape: {df_2017_prepared.shape}")
print(
    f"   Columns: {len(df_2017_prepared.columns)} (77 features + label + source)")

# Verify column alignment
print("\n✅ Verifying column alignment...")
cols_match = list(df_2017_prepared.columns) == list(df_2018_prepared.columns)
print(f"   Columns match: {cols_match}")

# Combine datasets
print("\n🔗 Concatenating datasets...")
df_unified = pd.concat(
    [df_2017_prepared, df_2018_prepared], axis=0, ignore_index=True)

print("\n✅ Unified dataset created!")
print(f"\n📏 Shape: {df_unified.shape}")
print(f"   - Total rows: {df_unified.shape[0]:,}")
print(f"   - Total columns: {df_unified.shape[1]}")
print("   - Features: 77")
print("   - Label: 1 (UnifiedLabel)")
print("   - Metadata: 1 (Dataset_Source)")

# Memory usage
memory_mb = df_unified.memory_usage(deep=True).sum() / 1024**2
print(f"\n💾 Memory Usage: {memory_mb:.2f} MB")

# Class distribution
print("\n📊 Unified Label Distribution:")
print(df_unified['UnifiedLabel'].value_counts().sort_index())

# Dataset source distribution
print("\n📊 Dataset Source Distribution:")
print(df_unified['Dataset_Source'].value_counts())

# Check missing values
missing_count = df_unified.isnull().sum().sum()
missing_pct = (missing_count /
               (df_unified.shape[0] * df_unified.shape[1])) * 100
print("\n🔍 Missing Values:")
print(f"   Total: {missing_count:,} ({missing_pct:.2f}%)")

# Show which columns have missing values
cols_with_missing = df_unified.isnull().sum()[df_unified.isnull().sum() > 0]
if len(cols_with_missing) > 0:
    print(f"\n   Columns with missing values ({len(cols_with_missing)}):")
    for col, count in cols_with_missing.items():
        pct = (count / len(df_unified)) * 100
        print(f"      {col}: {count:,} ({pct:.1f}%)")

print("\n✅ Dataset combination complete!")
```

```python
# Cell 8: Analyze Missing Features for Imputation Strategy
# Purpose: Examine distributions of 20 features to choose proper imputation method

print("🔍 Analyzing Missing Features for Imputation")
print("=" * 60)

# Get the 20 features with missing values
missing_features = additional_2018_features

print(f"\n📊 Analyzing {len(missing_features)} features with missing values...")
print("   (These exist only in 2018 data)\n")

# Analyze each feature in 2018 data (where they exist)
df_2018_only = df_unified[df_unified['Dataset_Source'] == '2018']

imputation_strategy = {}

for feature in missing_features:
    print(f"\n{'='*60}")
    print(f"Feature: {feature}")
    print(f"{'='*60}")

    # Get statistics from 2018 data
    values = df_2018_only[feature]

    print(f"Data type: {values.dtype}")
    print(f"Non-null count: {values.notna().sum():,}")
    print("Statistics:")
    print(f"   Min:    {values.min()}")
    print(f"   Max:    {values.max()}")
    print(f"   Mean:   {values.mean():.4f}")
    print(f"   Median: {values.median():.4f}")
    print(f"   Std:    {values.std():.4f}")

    # Count unique values
    unique_count = values.nunique()
    print(f"   Unique values: {unique_count}")

    # Check if binary or categorical
    if unique_count <= 10:
        print("   Value distribution:")
        print(values.value_counts().head(10))

    # Determine imputation strategy
    if 'Flag' in feature or 'Count' in feature:
        # Flag counts - typically 0 or small integers
        strategy = 'zero'  # Most flows don't have special flags
        print("   → Imputation: Fill with 0 (flags typically absent)")
    elif 'Bulk' in feature:
        # Bulk transfer metrics - often 0 for non-bulk transfers
        strategy = 'zero'
        print("   → Imputation: Fill with 0 (most flows aren't bulk)")
    elif 'Protocol' in feature:
        # Protocol number - categorical
        strategy = 'mode'  # Most common protocol
        mode_val = values.mode()[0]
        print(f"   → Imputation: Fill with mode = {mode_val}")
    elif 'Min' in feature:
        # Minimum packet lengths
        strategy = 'median'
        print(f"   → Imputation: Fill with median = {values.median():.2f}")
    elif 'Ratio' in feature:
        # Ratios
        strategy = 'median'
        print(f"   → Imputation: Fill with median = {values.median():.2f}")
    else:
        # Default: median
        strategy = 'median'
        print(f"   → Imputation: Fill with median = {values.median():.2f}")

    imputation_strategy[feature] = {
        'method': strategy,
        'value': values.mode()[0] if strategy == 'mode' else values.median() if strategy == 'median' else 0
    }

print(f"\n\n{'='*60}")
print("📋 IMPUTATION STRATEGY SUMMARY")
print(f"{'='*60}")

for feature, strategy in imputation_strategy.items():
    print(
        f"{feature:30s} → {strategy['method']:10s} (value: {strategy['value']})")

print("\n✅ Imputation strategy defined")
```

```python
# Cell 9: Apply Imputation
# Purpose: Fill missing values using defined strategy

print("🔧 Applying Imputation to Missing Features")
print("=" * 60)

# Make a copy to preserve original
df_unified_imputed = df_unified.copy()

print("\n📊 Before Imputation:")
print(f"   Total missing values: {df_unified_imputed.isnull().sum().sum():,}")

# Apply imputation based on strategy
print("\n🔄 Applying imputation...")

for feature, strategy_info in imputation_strategy.items():
    method = strategy_info['method']
    value = strategy_info['value']

    # Count missing before
    missing_before = df_unified_imputed[feature].isnull().sum()

    # Apply imputation
    df_unified_imputed[feature].fillna(value, inplace=True)

    # Count missing after (should be 0)
    missing_after = df_unified_imputed[feature].isnull().sum()

    print(
        f"   {feature:30s}: {missing_before:,} → {missing_after:,} (filled with {value})")

print("\n📊 After Imputation:")
print(f"   Total missing values: {df_unified_imputed.isnull().sum().sum():,}")

# Verify no missing values remain
if df_unified_imputed.isnull().sum().sum() == 0:
    print("\n✅ All missing values successfully imputed!")
else:
    print("\n⚠️  WARNING: Some missing values remain!")
    remaining = df_unified_imputed.isnull().sum(
    )[df_unified_imputed.isnull().sum() > 0]
    print(remaining)

# Final dataset info
print("\n📏 Final Unified Dataset:")
print(f"   Shape: {df_unified_imputed.shape}")
print(f"   Rows: {df_unified_imputed.shape[0]:,}")
print(f"   Columns: {df_unified_imputed.shape[1]}")
print(
    f"   Memory: {df_unified_imputed.memory_usage(deep=True).sum() / 1024**2:.2f} MB")

print("\n📊 Label Distribution:")
print(df_unified_imputed['UnifiedLabel'].value_counts())

print("\n✅ Imputation complete - unified dataset ready!")
```

```python
# Cell 10: Save Unified Dataset
# Purpose: Save processed unified dataset to CSV

print("💾 Saving Unified Dataset")
print("=" * 60)

# Define output path
output_path = PROCESSED_DIR / 'cic_unified_dataset.csv'

print("\n📁 Output path:")
print(f"   {output_path}")

# Save to CSV
print("\n⏳ Saving dataset (this may take a few minutes for 15.8M rows)...")
df_unified_imputed.to_csv(output_path, index=False)

# Verify file was created
file_size_mb = output_path.stat().st_size / 1024**2
print("\n✅ Dataset saved successfully!")
print(f"   File size: {file_size_mb:.2f} MB")
print(f"   Location: {output_path}")

# Save summary statistics
summary = {
    'total_rows': len(df_unified_imputed),
    'total_columns': len(df_unified_imputed.columns),
    'features': 77,
    'label_column': 'UnifiedLabel',
    'num_classes': df_unified_imputed['UnifiedLabel'].nunique(),
    'class_distribution': df_unified_imputed['UnifiedLabel'].value_counts().to_dict(),
    'dataset_sources': df_unified_imputed['Dataset_Source'].value_counts().to_dict(),
    'file_size_mb': file_size_mb
}

print("\n📊 Dataset Summary:")
print(f"   Total samples: {summary['total_rows']:,}")
print(f"   Features: {summary['features']}")
print(f"   Classes: {summary['num_classes']}")
print("   Source breakdown:")
for source, count in summary['dataset_sources'].items():
    pct = (count / summary['total_rows']) * 100
    print(f"      {source}: {count:,} ({pct:.1f}%)")

print("\n✅ Unified dataset ready for model training!")
print("\n📋 Next steps:")
print("   1. Move to notebook 02_feature_engineering.ipynb")
print("   2. Or directly to 03_model_training.ipynb")
```
